```python
import pandas as pd

# Cargar el CSV de Eurostat
df_raw = pd.read_csv("../data/csv/tec00118_linear.csv")

# Ver las primeras filas para entender la estructura
print("🔎 Estructura original:")
print(df_raw.head(10))

# Normalizar nombres de columnas
df_raw.columns = [c.strip().lower() for c in df_raw.columns]

# En los CSV de Eurostat, la columna con fechas suele llamarse 'TIME' o 'TIME_PERIOD'
# y la columna con los valores suele llamarse 'Value' o 'OBS_VALUE'
fecha_col = [c for c in df_raw.columns if "time" in c][0]
valor_col = [c for c in df_raw.columns if "value" in c or "obs" in c][0]

# Crear DataFrame limpio
df_infl = df_raw[[fecha_col, valor_col]].copy()
df_infl.columns = ["date", "value"]

# Convertir fechas y valores numéricos
df_infl["date"] = pd.to_datetime(df_infl["date"], errors="coerce")
df_infl["value"] = pd.to_numeric(df_infl["value"], errors="coerce")

# Eliminar filas vacías y ordenar
df_infl = df_infl.dropna().sort_values("date").reset_index(drop=True)

print("\n✅ Inflación limpia:")
print(df_infl.head())

# Guardar lista para integración posterior
df_infl.to_csv("../data/csv/euro_inflation_clean.csv", index=False)
print("\n💾 Guardado como '../data/csv/euro_inflation_clean.csv'")
```
