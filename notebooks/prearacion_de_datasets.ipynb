```python
import pandas as pd
df_full = pd.read_csv("../data/csv/raw/BBVA_features.csv", parse_dates=["Date"])
# 🔹 Supongamos que tu DataFrame final se llama df_full
# y tiene una columna 'date' como índice o columna
df_full = df_full.copy()
df_full["eventos_negativos"] = 0  # Por defecto, 0

# =====================================================
# 🧱 1️⃣ Crisis financiera global (2008 - 2009)
# =====================================================
mask_2008 = (df_full["Date"] >= "2008-09-01") & (df_full["Date"] <= "2009-06-30")
df_full.loc[mask_2008, "eventos_negativos"] = 1

# =====================================================
# 🦠 2️⃣ Pandemia COVID-19 (marzo 2020 - junio 2020)
# =====================================================
mask_covid = (df_full["Date"] >= "2020-03-01") & (df_full["Date"] <= "2020-06-30")
df_full.loc[mask_covid, "eventos_negativos"] = 1

# =====================================================
# 💣 3️⃣ Guerra Rusia - Ucrania (febrero 2022 - septiembre 2022)
# =====================================================
mask_ukraine = (df_full["Date"] >= "2022-02-24") & (df_full["Date"] <= "2022-09-30")
df_full.loc[mask_ukraine, "eventos_negativos"] = 1

# =====================================================
# 📉 4️⃣ Burbuja tecnológica (Dot-com crash, 2000 - 2001)
# =====================================================
mask_dotcom = (df_full["Date"] >= "2000-03-01") & (df_full["Date"] <= "2001-03-31")
df_full.loc[mask_dotcom, "eventos_negativos"] = 1

# =====================================================
# 💥 5️⃣ Crisis de deuda europea (julio 2011 - diciembre 2012)
# =====================================================
mask_deuda = (df_full["Date"] >= "2011-07-01") & (df_full["Date"] <= "2012-12-31")
df_full.loc[mask_deuda, "eventos_negativos"] = 1

# =====================================================
# 🏦 6️⃣ Colapso del Silicon Valley Bank (marzo 2023)
# =====================================================
mask_svb = (df_full["Date"] >= "2023-03-10") & (df_full["Date"] <= "2023-04-15")
df_full.loc[mask_svb, "eventos_negativos"] = 1

event_periods = [
    ("2001-09-11", "2001-10-31"),
    ("2000-03-01", "2001-03-31"),
    ("2011-07-01", "2012-12-31"),
    ("2010-05-06", "2010-05-10"),
    ("2008-09-01", "2009-06-30"),
    ("2020-03-01", "2020-06-30"),
    ("2022-02-24", "2022-09-30"),
]

for start, end in event_periods:
    mask = (df_full["Date"] >= start) & (df_full["Date"] <= end)
    df_full.loc[mask, "eventos_negativos"] = 1

# =====================================================
# 🔹 Verificación rápida
# =====================================================
print(df_full["eventos_negativos"].value_counts())
print(df_full[df_full["eventos_negativos"] == 1].head(10))

# =====================================================
# 💾 Guardar versión extendida
# =====================================================
df_full.to_csv("../data/csv/clean/BBVA_macro_extended.csv", index=False)
print("✅ Archivo guardado con columna 'eventos_negativos' añadida.")
```

```python
import yfinance as yf
import pandas as pd

# 1️⃣ Descargar dividendos de BBVA (formato oficial de Yahoo Finance)
bbva = yf.Ticker("BBVA.MC")
div_bbva = bbva.dividends.rename("div_amount").to_frame()
santander = yf.Ticker("SAN.MC")
div_san = santander.dividends.rename("div_amount").to_frame()

# 2️⃣ Convertir índice a columna de fecha
div_bbva.reset_index(inplace=True)
div_bbva.rename(columns={"Date": "date"}, inplace=True)
div_san.reset_index(inplace=True)
div_san.rename(columns={"Date": "date"}, inplace=True)

# 3️⃣ Guardar en CSV
ruta_salida = "../data/csv/raw/dividends_BBVA.csv"
div_bbva.to_csv(ruta_salida, index=False)
ruta_salida_san = "../data/csv/raw/dividends_SAN.csv"
div_san.to_csv(ruta_salida_san, index=False)

print(f"✅ Dividendos guardados en: {ruta_salida}")
print(div_bbva.head())
```
